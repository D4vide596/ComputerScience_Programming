# Recursive Function

Factorial Recursive Function

```python
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n-1)
```

Incorrect recursion example

```python
def bad_factorial(n):
    if n == 0:
        return 1
    else:
        return n * bad_factorial(n)
```

Exercise, mistery function

```python
def mystery(n):
    if n <= 1:
        return n
    else:
        return mystery(n-1) + mystery(n-2)
```

Merge Sort recursive — Python implementation

```python
def merge_sort(arr):
    if len(arr) <= 1:
        return arr
    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])
    return merge(left, right)
```

Merge Sort — Python implementation

```python
def merge(left, right):
    result = []
    i=j=0
    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    result.extend(left[i:])
    result.extend(right[j:])
    return result
```

Given the list: [1, [2, 3], [4, [5, 6]], 7]
Write a recursive function: def deep_sum(data)
that returns the sum of all numbers in the list.

```python
def deep_sum(data):
    total = 0
    for x in data:
        if isinstance(x, list):
            total += deep_sum(x)
        else:
            total += x
    return total
```

* Example problem: generating subsets
* Example set: [1, 2, 3]
* Subsets:
* * []
* * [1]
* * [2]
* * [3]
* * [1, 2]
* * [1, 3]
* * [2, 3]
* * [1, 2, 3]
* For a set of size n:
* Number of subsets:
* * 2n
*
* Each element creates two choices:
* * exclude the element
* * include the element
* We will solve this problem using:
* * recursion
* * iteration

Recursive Solution

```python
def generate_subsets(nums):
    if not nums:
        return [[]]
    subsets = generate_subsets(nums[:-1]) # remove last element new_subsets = []
    for s in subsets:
        new_subsets.append(s + [nums[-1]])
    return subsets + new_subsets
```

Iterative solution

```python
def generate_subsets_iter(nums):
    subsets = [[]]
    for x in nums:
        new_sets = []
        for s in subsets: # Each iteration doubles the number of subsets
            new_sets.append(s + [x])
        subsets = subsets + new_sets
    return subsets
```
