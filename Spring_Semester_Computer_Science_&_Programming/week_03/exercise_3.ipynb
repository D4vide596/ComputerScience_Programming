# Exercise 3 — Mixed Linear and Logarithmic Growth

Consider:

```python
def strange(n):
    total = 0
    i=1
    while i < n:
        for j in range(n):
            total += 1
        i *= 2
    return total
```

1. How many times does the while loop execute?

$
\begin{gathered}
2^x = n \\
\log_{2}(n)
\end{gathered}
$

2. How many times does the inner statement execute overall?

$n \log_2(n)$

3. Give the asymptotic upper bound.

$T(n) \in O(n \log_2 n)$

4. Compare this complexity to O(n2). Which grows faster?

the $O(n \log_2 n)$ is faster than the $O(n^2)$
