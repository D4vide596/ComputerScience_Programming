# Exercise 2 — Early Termination and Case Analysis

Consider the function:

```python
def has_duplicate(data):
    iteration = 0
    n = len(data)
    for i in range(n):
        for j in range(i+1, n):
            if data[i] == data[j]:
                return iteration
            iteration += 1
        iteration += 1
    return False
```

1. What is the worst-case time complexity?

that we dont have any duplicate or the duplicate are the last two elements

2. What is the best-case time complexity?

that the first two elements are duplicate

5. Suggest conceptually how this algorithm could be improved.

```python
def has_duplicate_better(data):
    iteration = 0
    data_sorted = sorted(data)
    n = len(data_sorted)
    for i in range(n-1):
        if data_sorted[i] == data_sorted[i+1]:
            return iteration
        iteration += 1
    return False
```
