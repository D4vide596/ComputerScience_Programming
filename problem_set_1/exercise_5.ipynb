Consider the following class:

```python
from course import Course
class Student:
    courses = []
    def add_course(self, course):
        self.courses.append(course)
```

1. Create two students.

```python
student_1 = Student()
student_2 = Student()
```

2. Add a course to only one.

```python
course_math = Course("math", 6)
student_1.add_course(course_math)
print(student_1.courses)
print(student_2.courses)
```

3. What happens?

- both of the class have the same course

4. Why does this happen?
- because we use a the list of course as a class variable instead of a object variable (dont remember the name sorry)
