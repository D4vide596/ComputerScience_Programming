# Exercise 4 — Student and Object Composition

Implement a class Student with:




Attributes
- name
- student_id
- a list of Course objects

Constraints
- A student cannot have two courses with the same name.

Methods
- add_course(course)
- total_ects_passed()
- weighted_average() (weighted by ECTS)
- \_\_str__()

```python
from course import Course
class Student:
    def __init__(self, name, student_id):
        self.name = name
        self.student_id = student_id
        self.courses = []

    def add_course(self, course):
        if course not in self.courses:
            self.courses.append(course)
            print("Course added in List")
        else:
            print("Course already in List")

    def total_ects_passed(self):
        sum_ects_passed = 0
        for course in self.courses:
            if course.is_passed():
                sum_ects_passed += course.ects
        return sum_ects_passed

    def weighted_average(self):
        try:
            sum_ects_per_grade = 0
            sum_ects = 0
            for course in self.courses:
                sum_ects_per_grade += course.ects * course.average()
                sum_ects += course.ects
            if sum_ects != 0:
                return round(sum_ects_per_grade/sum_ects, 2)
            else:
                raise ZeroDivisionError("The sum of ECTS is zero")
        except ZeroDivisionError as e:
            print(f"ZeroDivisionError: {e}")
            return 0

    def __str__(self):
        list_of_courses = [str(course) for course in self.courses]
        course_text = " | ".join(list_of_courses)
        return (f"Student: {self.name} (ID: {self.student_id})\n"
            f"Courses: [{course_text}]\n"
            f"Total ECTS: {self.total_ects_passed()}\n"
            f"Weighted Average: {self.weighted_average()}")
```

Then:

2. Add different courses to each.
3. Show that modifying one student does not affect the other.
4. Explain why this works.

```python
student_1 = Student("Davide", "01")
student_2 = Student("Elisa", "02")

math_std_1 = Course("math", 6)
math_std_2 = Course("math", 6)
italian_std_1 = Course("italian", 3)
italian_std_2 = Course("italian", 3)

student_1.add_course(italian_std_1)
student_1.add_course(math_std_1)

student_2.add_course(italian_std_2)
student_2.add_course(math_std_2)

math_std_1.add_grade(6.0)
math_std_2.add_grade(4.0)

physics_std_1 = Course("physics", 6)
physics_std_1.add_grade(5.0)
student_1.add_course(physics_std_1)

print(student_1)
print(student_2)

print(f"Identity check (Courses): {math_std_1 is math_std_2}")
print(f"Identity check (Lists): {student_1.courses is student_2.courses}")
```
