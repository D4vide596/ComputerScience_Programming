# Sample statistics

```python
import pandas as pd
from seaborn import load_dataset
# Load car_crashes dataset
car_crashes = load_dataset("car_crashes")
car_crashes.head()
```

```python
car_crashes.describe()
```

```python
numeric_columns = ['total', 'speeding', 'alcohol', 'not_distracted', 'no_previous',
       'ins_premium', 'ins_losses']
car_crashes[numeric_columns].quantile([0.01, 0.99])
```

# Histogram

```python
# plot an histogram of the alcohol column in the car_crashes dataset
import matplotlib.pyplot as plt
plt.figure(figsize=(6, 4))
car_crashes.alcohol.hist(bins=15)
plt.xlabel('% of crashes involving alcohol')
plt.ylabel('Counts')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

# Boxplot

```python
# plot a boxplot  of the alcohol and speeding columns in the car_crashes dataset
plt.figure(figsize=(6, 4))
car_crashes.boxplot(column=['alcohol', 'speeding'])
plt.ylabel('% of crashes involving...')
plt.show()
```

# QQplot

```python
# plot the QQplot of the alcohol column in the car_crashes dataset
import scipy.stats as stats
plt.figure(figsize=(6, 4))
stats.probplot(car_crashes['alcohol'], dist="norm", plot=plt)
plt.title('QQ Plot of Crashes Involving Alcohol')
plt.show()
```

# Bar plot

```python
import seaborn as sns 
# Load tips dataset
tips = load_dataset("tips")
tips.head()
```

```python
# Bar plot of the day column of tips dataset
plt.figure(figsize=(6, 4))
sns.countplot(x="day", data=tips, palette="Blues")
plt.xlabel('Day')
plt.ylabel('Count')
plt.title('Barplot of Day')
plt.show()
```
