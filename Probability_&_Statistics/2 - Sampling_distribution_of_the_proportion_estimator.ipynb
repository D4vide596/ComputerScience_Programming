# Sampling distribution of the proportion estimator
We consider a population of 10000 individuals of which 8800 support the expansion of solar energy.
We consider the statistic $p = $ (number of favorable outcomes)/n which gives an estimate of the proportion ($\pi$ = 0.88) of individuals supporting solar expansion based on a n-dimensional sample. We simulate 1000 different n-dimensional samples and compute the corresponding values of the statistic in order to analyse the distribution of the estimator $p$


```python
import numpy as np
support = np.ones(8800)
supportnot = np.zeros(1200)
# Build the population of 10000 individuals
population = np.append(support, supportnot)
```

The variables we are considering are $$
    X_i \begin{cases} 1 \quad & \text{if the $i$-th individual agrees with expansion of solar energy} \\ 0 &\text{otherwise} \end{cases}
    $$
    hence we can compute the amount of favorable outcomes and the population size.

```python
population.sum()
```

```python
len(population)
```

Usually, we work with just a sample of the entire population. The estimator for the sample mean is $p = \sum_{i=1}^{500} X_i$. 

```python
# Sample 500 individuals from the population
n = 500
sample = np.random.choice(population,n, replace=True)
# Estimate p from the sample
phat = sample.sum()/n
phat
```

In our experiment, we can draw several samples, computing $\hat{p}$ each time.

```python
# Repreat the above estimation 1000 times
phat_sample1 = np.array([])
for i in range(1000):
  sample = np.random.choice(population,n, replace=True)
  phat_sample1= np.append(phat_sample1, sample.sum()/n)
```

```python
# plot the distribution of the 1000 estimates obtained above
import matplotlib.pyplot as plt
plt.hist(phat_sample1, 20);
```

```python
phat_sample1.mean()
```

The sample mean of $\hat{P}$ is very close to the true parameter value. An estimator whose mean is equal to the true parameter value is said unbiased, since its bias (defines as $E[\hat{P}]-p$) is zero.

What happens if we estimate the proportion using only 100 samples?

```python
phat_sample2 = np.array([])
for i in range(1000):
  sample = np.random.choice(population,n, replace=True)
  phat_sample2= np.append(phat_sample2, sample[0:100].sum()/100)

print("estimator sample mean", phat_sample2.mean())
plt.hist(phat_sample2, 10);
```

```python
plt.hist(phat_sample1, 20, density=True)
plt.hist(phat_sample2, 20, density=True, alpha=0.5)
```

The first estimator (which uses 500 observations) has a smaller standard deviation than the second (using only 100 observations). The standard deviation of an estimator is called Standard Error and measures its precision. Estimators with a smaller standard error should be preferred.
