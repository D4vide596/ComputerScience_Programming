# Matplotlib exercises

## Import matplotlib (and numpy)

* Import ``matplotlib.pyplot`` using the standard name ``plt``
* Import ``numpy`` using the standard name ``np``

```python
import matplotlib.pyplot as plt
import numpy as np

print(plt.style.available)

plt.style.use('seaborn-v0_8-whitegrid')
plt.rcParams['figure.figsize'] = [8, 2.5]
plt.rcParams['font.size'] = 14
```

## Creating and saving plots

* Plot the function $y=\sin(x)$ on 100 points in the interval $[0, 2\pi]$ and save it to a file sinusoidal.png

```python
x = np.linspace(0, 2*np.pi, 100)
y = np.sin(x)

fig = plt.figure()
plt.plot(x, y)
plt.savefig("sinusoidal.png")
```

* Plot the functions $\sin(x)$ and $\cos(x)$ on 100 points in the interval $[0, 2\pi]$. 

```python
x = np.linspace(0, 2*np.pi, 100)
y1 = np.sin(x)
y2= np.cos(x)

fig = plt.figure()
plt.plot(x, y1)
plt.plot(x, y2)
```

*  Reproduce the plot below using matplotlib (with the same title, legend, gridlines, and style options)

<img src="img/trigonometric.png" aligh="left"></img>

```python
x = np.linspace(0, 2*np.pi, 100)
y1 = np.sin(x)
y2= np.cos(x)

plt.figure(figsize=[8,5])

plt.plot(x, y1, 'k--', label='sin(x)')
plt.plot(x, y2, 'b-.', label='cos(x)')
plt.ylabel('y')
plt.xlabel('x')
plt.title("Trionometric functions sin and cos")
plt.legend()
plt.grid(True)
```

## Subplots

*  Reproduce the plot below using matplotlib 

<img src="img/trigonometric_subplots.png" aligh="left"></img>

```python
x = np.linspace(0, 2*np.pi)
y1 = np.sin(x)
y2 = np.cos(x)

fig, ax = plt.subplots(2, sharey=True, figsize=(8, 6))
fig.suptitle('Trigonometric functions sin and cos')
ax[0].plot(x, y1, 'k--')
ax[0].set_ylabel('sin(x)')
ax[1].plot(x, y2, 'b-.')
ax[1].set_ylabel('cos(x)')
ax[1].set_xlabel('x')
```

## More advanced plots

Let us load the well-known iris dataset (csv file iris.csv)

```python
import pandas as pd
```

```python
iris = pd.read_csv("iris.csv") # load the dataset
iris.sample(5) # 5 random samples from the dataset
```

The dataset represent iris flowers belonging to three species: setosa, versicolor, virginica.



<img src="img/iris_species.png" aligh="left"></img>

For each specimen, 4 measurements are recorded:
 * sepal_length
 * sepal_width
 * petal_length
 * petal_width 

<img src="img/iris_petal_sepal.png" aligh="left"></img>


<br>
Our analysis is meant to answer the following questions:

1. Is there any relation between the 4 measurements and the species? 
2. Is it possible to guess the species, given the 4 measurements?

* Generate the scatterplot petal_width vs. petal_length. What do you observe?

```python
plt.figure(figsize=(10,6))
plt.plot(iris["petal_length"], iris["petal_width"], 'o');
```

* Generate the scatterplot petal_width vs. petal_length, with different colors for the 3 species. What do you observe?

```python
iris_set = iris[iris["species"] == "setosa"]
iris_ver = iris[iris["species"] == "versicolor"]
iris_vir = iris[iris["species"] == "virginica"]

plt.figure(figsize=(10,8))
plt.plot(iris_set["petal_length"], iris_set["petal_width"], 'ro', label="setosa");
plt.plot(iris_ver["petal_length"], iris_ver["petal_width"], 'go', label="versicolor");
plt.plot(iris_vir["petal_length"], iris_vir["petal_width"], 'bo', label="virginica");

plt.xlabel("Petal length (cm)")
plt.ylabel("Petal width (cm)")
plt.legend()
```
